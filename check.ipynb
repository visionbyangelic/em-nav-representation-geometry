```python
import gymnasium as gym
import minigrid
from minigrid.envs import EmptyEnv

# Bypassing pre-registered static IDs and instantiating a custom 12x12 matrix sandbox
env = EmptyEnv(size=12, render_mode=None)

# Reset the environment to configure the starting state
observation, info = env.reset(seed=42)

print("✅ MiniGrid Platform Custom 12x12 Sandbox Initialized Successfully!")
print("Initial Observation Keys Available:", observation.keys())
print("Grid Dimension Check:", f"{env.unwrapped.grid.width}x{env.unwrapped.grid.height}")
```
